```python
import pandas as pd

```

```python
df_270826 = pd.read_csv(
    "DVRS270826.txt",
    sep="|",
    encoding="utf-16",
    dtype=str
)


df_270826.columns = (
    df_270826.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

print(df_270826.head())
```
